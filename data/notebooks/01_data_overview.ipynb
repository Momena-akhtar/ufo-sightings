### Load datasets: Scrubbed Dataset

```python
scrubbed = pd.read_csv('../raw/scrubbed.csv', engine='python', on_bad_lines='skip')
scrubbed.head()
```

### Basic Structure

```python
scrubbed.shape
scrubbed.info()
scrubbed.describe(include='all')
```

```python
# Missing Values
scrubbed.isna().sum().sort_values(ascending=False)
```

```python
# Missing Values Pie Chart - Only columns with significant missing data
major_cols = ['datetime', 'city', 'state', 'country', 'latitude', 'longitude', 'duration (seconds)', 'shape', 'comments']
missing_counts = scrubbed[major_cols].isna().sum()

# Filter out columns with no or very few missing values
missing_counts = missing_counts[missing_counts > 0]

# Sort for better visualization
missing_counts = missing_counts.sort_values(ascending=False)

plt.figure(figsize=(10, 7))
colors = plt.cm.Set3(range(len(missing_counts)))
plt.pie(missing_counts, labels=missing_counts.index, autopct='%1.1f%%', 
        startangle=90, colors=colors)
plt.title('Missing Values Distribution', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Date time field quality
scrubbed['datetime'].head(20)
```

```python
# Parsing datetime field
pd.to_datetime(scrubbed['datetime'], errors='coerce').isna().mean()
```

```python
# Geographic Fields
scrubbed[['city','state','country','latitude']].sample(20)
```

```python
# Shape Column values
scrubbed['shape'].value_counts(dropna=False)
```

```python
# Comments Length Distribution
scrubbed['comments'].str.len().describe()
```
